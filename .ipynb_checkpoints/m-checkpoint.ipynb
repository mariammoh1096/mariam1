```python
import sqlite3
conn = sqlite3.connect('data.sqlite')
```

```python
cur = conn.cursor()
```

```python
cur.execute('''CREATE TABLE Counts
            (org TEXT, count INTEGER)''')
```

```python
cur.execute("INSERT INTO Counts(org, count) VALUES ('GAZA', 100)")
conn.commit()
```

```python
cur.execute('SELECT org, count FROM Counts')
for row in cur:
    print(row[0], row[1])
```

```python
cur.close()
conn.close()
```

```python
import sqlite3

conn = sqlite3.connect('music.sqlite')
cur = conn.cursor()

cur.execute('DROP TABLE IF EXISTS Track')
cur.execute('CREATE TABLE Track (title TEXT, plays INTEGER)')

conn.close()
```
